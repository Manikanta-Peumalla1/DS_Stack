# Decomposition Techniques — Overview

This notebook demonstrates decomposition methods (PCA, SVD, ICA) applied to the red-wine dataset for dimensionality reduction and feature analysis.
Key steps: data preparation, scaling, fit decomposition models, and visualize components/variance explained.

Notes: check scaling and missing values before fitting PCA/SVD.


```python
# Importing required libraries for data manipulation, visualization, and database connection
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
import os
import dotenv
import warnings
# Suppress warnings for cleaner output
warnings.filterwarnings(action='ignore')
```

```python
# Establishing connection to MySQL database using credentials from environment variables
engine = create_engine("mysql+pymysql://{user}:{pw}@localhost/{db}"
                       .format(user = os.getenv('USER'),    # user
                               pw = os.getenv('PASSWORD'),      # password
                               db = os.getenv('DB_NAME'))) # database
```

```python
# Querying the wine quality data from the database and saving it as a CSV file
sql='select * from mission2026.`winequality-red`;'  # SQL query to fetch data
# Read data from SQL into a DataFrame
data=pd.read_sql(sql=sql,con=engine)
# Save the DataFrame to a CSV file
with open('Data/red-wine.csv','wb') as f:
    data.to_csv(f)
    
# Make a copy of the data for further processing
df=data.copy()
```

```python
# Apply Principal Component Analysis (PCA) to reduce dimensionality to 5 components
from sklearn.decomposition import PCA
pca = PCA(n_components=5)
# Fit and transform the data using PCA
pca.fit_transform(df)
```
